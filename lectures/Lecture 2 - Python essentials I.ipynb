## Mutability

Every variable in Python is an object. An object consists of two things:

- reference: indicates where (i.e. memory address) you can find the actual value
- value: the actual value you assign to an object

`id()` in Python shows the reference information of an object.

Immutable objects cannot change the value once defined. This means that if you change the value of an immutable object, you are actually creating a new object and assigning it to the same reference box.

Mutable objects can change the value once defined. This means that if you change the value of a mutable object, you are actually changing the value of the same object.

````python
x = 5
print(id(x))
x = 6
print(id(x))
````

Question: So is an integer mutable or immutable?

How about a list?

````python
x = [1, 2, 3]
print(id(x))
x.append(4)
print(id(x))
````

In Python, integers, floats, strings, and tuples are immutable, while lists, sets, and dictionaries are mutable.

## Dictionaries

Dictionaries store key-value pairs:

- The key is a unique identifier used to find the value.
- The value is the data associated with that key.

````python
student = {'name': "Veena", 'age': 18, 'courses': ["dsc10", "dsc20"]}
print(student)
````

Keys must be of an immutable data type, while values can be of any type.

Otherwise, keys could be changed, and the dictionary would not be able to find the value associated with the key!

### Retrieving Values

`get(<key>)` and `<dict>[<key>]` both return the corresponding value if the key exists.

However, `get(<key>)` returns `None` if the key does not exist, while `<dict>[<key>]` will throw an error.

````python
print(student.get('name'))  # Output: Veena
print(student.get('grade'))  # Output: None
print(student.get('grade', 'n/a'))  # Output: n/a
print(student['age'])  # Output: 18
````

Be careful with `get` though. It never fails, so a typo in a key name gives you `None` rather than an error, and that `None` can travel quite a long way before anything goes wrong.

## Sets

Sets are unordered collections of unique elements. They are useful for:
- Removing duplicates from a list.
- Checking for membership.
- Performing set operations like union, intersection and difference.

````python
my_set = {1, 2, 3, 4, 5, 5}
print(my_set)  # Output: {1, 2, 3, 4, 5}

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(set1 | set2)  # union: {1, 2, 3, 4, 5, 6, 7, 8}
print(set1 & set2)  # intersection: {4, 5}
print(set1 - set2)  # difference: {1, 2, 3}
````

Checking whether something is `in` a set is also much faster than checking whether it is `in` a list. We'll see why when we get to hash tables in Module C.

## Control Flow: Conditionals and Loops

Control flow allows for the manipulation of the order in which code is executed.

Conditionals control which code block is executed based on specific conditions using `if`, `else`, and `elif`.

````python
if x > y:
    print("x is greater than y")
else:
    print("x is not greater than y")
````

### Loops

Loops allow for repetitive tasks without writing redundant code.

**`for` loop**: repeats for a known (but unspecified) number of iterations.

````python
my_list = [0, 4, 0, 6, 5]
for u in my_list:
    print(u)
````

If you ever need the loop index, you can use the `enumerate()` function:

````python
for i, item in enumerate(my_list):
    print(i, item)
````

**`while` loop**: repeats while a condition is `True`.

````python
total = 0
n = 1
while total < 20:
    total += n
    n += 1
print(n, total)
````

### Break and Continue

### Parameters vs Arguments

Parameters are the inputs defined by the function, while arguments are the actual values you pass in when calling the function.

````python
def discriminant(a, b, c):
    return b**2 - 4*a*c

# Function call with arguments
discriminant(1, 2, 3)
````

We can provide default values for parameters, which allows us to call the function without providing that parameter.

It is also possible to pass arguments by name, which allows you to pass arguments in any order:

````python
def discriminant(a=0, b=0, c=0):
    return b**2 - 4*a*c

print(discriminant(1, 3))
print(discriminant(c=1, b=2))
````

It's good practice to provide a docstring for your functions, which describes what the function does and what parameters it expects.

````python
def discriminant(a=0, b=0, c=0):
    """
    This function returns the discriminant of a quadratic equation.

    Parameters:
    a (int): Coefficient of x^2, default is 0
    b (int): Coefficient of x, default is 0
    c (int): Constant, default is 0
    """
    return b**2 - 4*a*c

help(discriminant)
````

Agents are generally very good at writing docstrings. Do read them, but check them against what the code actually does. They don't always agree!

### Quick quiz: spot the bug

Here's a function to collect temperature readings from a weather station. What does the last line print?

 - A. `[19.0]`
 - B. `[21.5, 19.0]`
 - C. `[]`
 - D. An error

````python
def add_reading(value, readings=[]):
    readings.append(value)
    return readings

station_a = add_reading(21.5)
station_b = add_reading(19.0)
print(station_b)
````

Default values are evaluated once, when the function is defined, not each time it is called. So every call that uses the default shares **the same list**.

This is mutability again, and it's a nasty one: the code runs without any errors, and a test that only calls it once will pass.

The usual fix is to use `None` as the default:

````python
def add_reading(value, readings=None):
    if readings is None:
        readings = []
    readings.append(value)
    return readings

station_a = add_reading(21.5)
station_b = add_reading(19.0)
print(station_b)
````

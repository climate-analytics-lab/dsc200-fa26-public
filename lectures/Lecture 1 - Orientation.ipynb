# Why this course is different this year

Last year, DSC 200 spent most of the quarter on Python and the PyData libraries,
and the syllabus said:

> *"Avoid using generative AI (e.g., ChatGPT) in assignments or projects."*

This year that line is **inverted**. Most working code is now produced by
steering an agent, not by hand-writing library calls.

A course that trains you to hand-write `df.groupby(...)` from memory is training
you for the part of the job that is being automated.

## So what is actually scarce now?

Not the typing. The **judgment around the code**:

- turning a vague request into a precise **specification**
- choosing **data structures and algorithms** with the right trade-offs
- **validating** that generated code — and its output — is actually correct
- knowing when a thing is **done**

That is what this course teaches.

You will still learn Python, NumPy, pandas, xarray and matplotlib — but to the
level where you can **read, verify and steer**, not recall from memory.

## Module C deserves a word

Six sessions on complexity, arrays, hash tables, stacks, trees, heaps, sorting, graphs.

You might reasonably ask why, in 2026, when an agent will write the heap for you.

Because the agent picks these **constantly and silently**, usually behind a
library call — and code that is correct on your 10,000-row sample can be
quadratic in time or memory at full scale.

You are not going to implement a red-black tree. You are going to learn to
**recognise what got chosen, predict how it scales, and measure it.**

## The setup

I have a CSV of 10,000 NYC yellow-taxi trips from 2019.

I am going to give Claude Code a request of the kind you would genuinely get
from a stakeholder — which is to say, an underspecified one:

> *"Here's a CSV of NYC taxi trips. Which pickup zones tip the best?
> Give me the top 5."*

**Before I run it — poll `lec1-e`:**

*What do you expect to go wrong?*

1. It will crash / write code that doesn't run
2. It will run, and the answer will be right
3. It will run, and the answer will be wrong in a way I can see
4. It will run, and the answer will be wrong in a way I **can't** see

*Switching to the terminal — let's watch it work.*

## What it gave me

Reasonable-looking code. Runs clean, no warnings.

```python
import pandas as pd

df = pd.read_csv("../data/2019_Yellow_Taxi_Trip_Data.csv")

# Tip as a percentage of the total fare
df["tip_pct"] = df["tip_amount"] / df["total_amount"] * 100

top5 = (
    df.groupby("pulocationid")["tip_pct"]
      .mean()
      .sort_values(ascending=False)
      .head(5)
)
print(top5)
```

## Is that an answer you would send to your stakeholder?

### Bug #1 — the one you can see

Let's just ask how many trips are behind each of those numbers.

```python
counts = df.groupby("pulocationid").size()
pd.DataFrame({"tip_pct": top5, "n_trips": counts.reindex(top5.index)})
```

An 89% tip — from a **single trip**.

The agent was never told there should be a minimum sample size, so there wasn't one.

### Bug #2 — the one you can't

Fix the sample size and the numbers look sane. So we're done, right?

```python
busy = counts[counts >= 100].index
fixed = (
    df[df["pulocationid"].isin(busy)]
      .groupby("pulocationid")["tip_pct"].mean()
      .sort_values(ascending=False).head(5)
)
print(fixed)
```

Plausible. Defensible. Roughly 11–12%, which is about what you'd expect a tip to be.

And still wrong.

Look at what `payment_type` does to `tip_amount`:

```python
df.groupby("payment_type")["tip_amount"].agg(["count", "mean", "max"])
```

`payment_type` 2 is **cash** — about 28% of all trips — and the meter records
cash tips as `0.00`. They aren't zero. They're *unknown*.

Every zone's average is being dragged down by a column of structural zeros.

So what happens if we ask the question the stakeholder actually meant —
*of the trips where we can observe a tip, what fraction of the fare was it?*

```python
card = df[(df["payment_type"] == 1) & (df["pulocationid"].isin(busy))].copy()
card["tip_pct"] = card["tip_amount"] / card["fare_amount"] * 100

corrected = card.groupby("pulocationid")["tip_pct"].mean().sort_values(ascending=False)
pd.DataFrame({"agent": fixed, "corrected": corrected.reindex(fixed.index)})
```

```python
print("agent's top 5:     ", list(fixed.index))
print("corrected top 5:   ", list(corrected.head(5).index))
```

## The point

- The average roughly **doubles**: ~11% → ~27%
- And the top 5 is a **different set of zones**, in a different order

Nothing crashed. Nothing warned. The first answer looked completely reasonable,
and you would only catch it if you knew how the meter records a cash tip.

### None of these are the agent's fault, exactly

It was asked a vague question and it answered it. It had no access to the data
dictionary, and no one told it what "tips the best" meant.

- *"tip the best"* → **specification** (Module B, Module E)
- *what does `payment_type = 2` mean?* → **reading and verifying** (Module A)
- *how would I have caught this?* → **testing and validation** (Module D)
- *and what happens to all of this at 10⁸ rows?* → **Module C**

That is the whole course, in one wrong table.
